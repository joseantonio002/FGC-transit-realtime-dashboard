```python
import duckdb
```

```python
trips = duckdb.read_csv('./20260211_090010_TITSA_Tenerife/trips.txt')
```

```python
calendar_dates = duckdb.read_csv('./20260211_090010_TITSA_Tenerife/calendar_dates.txt').project("""
    *,
    STRPTIME(date::VARCHAR, '%Y%m%d')::DATE AS real_date
""")
```

```python
duckdb.query("""SELECT * 
                FROM trips 
                INNER JOIN 
                calendar_dates 
                ON trips.service_id = calendar_dates.service_id
                ORDER BY real_date DESC""")
```

```python
duckdb.query("""SELECT * 
                FROM trips 
                INNER JOIN 
                calendar_dates 
                ON trips.service_id = calendar_dates.service_id
                ORDER BY real_date ASC""")
```
